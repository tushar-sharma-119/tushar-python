```python
# filter
import numpy as np
arr = np.array([10,20,3,4,90,200])
arr_f = arr[arr>20] # filter
print(arr_f)
```

```python
import numpy as np
arr = np.array([10,20,30,6,3,2])
arr_f = arr[arr!=6] # filter
print(arr_f)
```

```python
arr = np.array([10,20,30,15,17,6,3,2])
arr_f= arr[(arr>10) & (arr<20)]
print(arr_f)
```

```python
#fancing indexing
import numpy as np
arr = np.array([10,20,30,3,4,90,100])
print(arr)
arr_f = arr[[0,2]]
print(arr_f)
```

```python
import numpy as np
arr = np.array([10,20,30,3,4,90,100])
print(arr)
arr_w = np.where(arr>9,"true","false")
print(arr_w)
```

```python
arr = np.array([10,20,30,3,4,90,100])
print(arr)
arr_w = np.where((arr>10)&(arr>20),arr*2,arr+2)
print(arr_w)
```

```python
# data structure: [restaurent_id'2021,2022,2023,2024]
import numpy as np

sales_data = np.array([
    [1,150000,180000,220000,250000],# paradise biryani
    [2,120000,140000,160000,190000],# beijing bites
    [3,200000,230000,260000,300000],# pizza hub
    [4,180000,210000,240000,270000],# buger point
    [5,160000,185000,205000,230000],# chai point
])

print(sales_data)
print(sales_data.shape)
print("dimension:",sales_data.ndim)
print("total elements:",sales_data.size)
print("restaurent_id:",sales_data[:,0])
print("2021 sales:",sales_data[:,1])
print("2024 sales:",sales_data[:,4])
print(np.max(sales_data[:,1:]))
print(np.min(sales_data[:,1:]))
print("\ntotal 2021 sales:",np.sum(sales_data[:,1]))
print("\ntotal 2024 sales:",np.mean(sales_data[:,4]))
print(np.sum(sales_data[:,1:], axis=1))
print(np.mean(sales_data[:,1:], axis=1))
```
